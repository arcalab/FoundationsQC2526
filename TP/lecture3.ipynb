```python
import pennylane as qml 
```

```python
n_qubits = 2 

dev = qml.device("default.qubit", wires=n_qubits)
```

```python


@qml.qnode(dev)
def qc():
    qml.Hadamard(wires=0)
    qml.Hadamard(wires=1)

    return qml.probs(wires=[0,1]) # = qml.probs() 


```

```python
qc() 
```

```python
qml.draw_mpl(qc)() 
```

```python

n_qubits = 2

dev = qml.device("default.qubit" , wires=n_qubits)

@qml.qnode(dev)
def qc():

    qml.Hadamard(wires=0)

    #passar qubit que esta na posição 1 para o estado |1>
    qml.PauliX(wires=1)

    qml.Hadamard(wires=1)
    qml.PauliZ(wires=1)
    qml.Hadamard(wires=1)
    



    #return qml.probs(wires=[0,1]) # = qml.probs() 
    # também podemos retornar o estado , ou seja, as amplitudes (complexas)
    return qml.state()

```

```python
qc()
```

```python
qml.draw_mpl(qc)()
```
