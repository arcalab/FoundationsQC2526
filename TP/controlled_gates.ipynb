```python
import pennylane as qml
```

```python
n_qubits = 2 

dev = qml.device("default.qubit" , wires=n_qubits)

@qml.qnode(dev)
def qc():
    #começamos com |00> 

    #preparar o estado de sobreposição uniforme
    qml.Hadamard(wires=0)
    qml.Hadamard(wires=1)

    #aplicar controlled Z 

    qml.CZ(wires=[0,1])

    qml.PauliX(wires=0)
    qml.CZ(wires=[0,1])
    qml.PauliX(wires=0)

    return qml.state()

```

```python
qc()
```

```python
qml.draw_mpl(qc)()
```

```python
n_qubits = 3 

dev = qml.device("default.qubit" , wires=n_qubits)

@qml.qnode(dev)
def qc():
    #começamos com |000> 

    #preparar o estado de sobreposição uniforme
    #qml.Hadamard(wires=0)
   #qml.Hadamard(wires=1)
    #qml.Hadamard(wires=1)
    for i in range(n_qubits):
        qml.Hadamard(wires=i)

    #aplicar Toffoli -> qubits na posição 0 ,1 são controlos e qubit em 2 é o target ou seja sofre o bit flip

    qml.Toffoli(wires=[0,1,2])

    qml.PauliX(wires=1)
    qml.Toffoli(wires=[0,1,2])
    qml.PauliX(wires=1)
    return qml.state()

```

```python
qml.draw_mpl(qc)()
```
