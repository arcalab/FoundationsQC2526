```python
import pennylane as qml
import numpy as np
```

```python
dev = qml.device("default.qubit", wires=2)

def oracle(id):
    if id==0:
        pass
    elif id==1:
        qml.PauliX(wires=1)
    elif id==2:
        qml.CNOT(wires=[0,1])
    elif id==3:
        qml.PauliX(wires=0)
        qml.CNOT(wires=[0,1])
        qml.PauliX(wires=0)


@qml.qnode(dev)
def deutsch(id):
    # inputs da função em sobreposição 
    qml.Hadamard(wires=0)

    #qubit auxiliar que guarda resultado da função 
    qml.PauliX(wires=1)
    qml.Hadamard(wires=1)

 

    #função 
    oracle(id)

    qml.Barrier()

    #interferencia
    qml.Hadamard(wires=0)


    return qml.probs(wires=0)
```

```python
id=2
qml.draw_mpl(deutsch)(id)
```
