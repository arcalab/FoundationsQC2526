```python
import pennylane as qml
import numpy as np
```

```python
dev = qml.device("default.qubit", wires=2)

@qml.qnode(dev)
def qc():
    # |psi>
    qml.RY(2*np.arccos(3/5), wires=0) #qml.PauliRot(2*np.arccos(3/5), "Y" , wires=0)
    qml.PauliZ(wires=0)
    qml.PhaseShift(np.pi/6, wires=0)

    # |-> 
    qml.PauliX(wires=1)
    qml.Hadamard(wires=1)

    return qml.probs()
```

```python
qml.draw_mpl(qc)()
```

```python
probabilidades = qc()
```

```python
probabilidades
```

```python
dev = qml.device("default.qubit", wires=2)

@qml.qnode(dev)
def qc():
    # Bell state 
    qml.Hadamard(wires=0)
    qml.CNOT(wires=[0,1])

    return qml.probs()
```

```python
qml.draw_mpl(qc)()
```

```python
qc()
```

```python
dev = qml.device("default.qubit", wires=3)

@qml.qnode(dev)
def qc():
    # GHZ state 
    qml.Hadamard(wires=0)
    qml.CNOT(wires=[0,1])
    qml.CNOT(wires=[0,2]) # igual  qml.CNOT(wires=[1,2]) tambem prepara GHZ

    return qml.probs()
```

```python
qml.draw_mpl(qc)()
```

```python
# ghz com 10 qubits


dev = qml.device("default.qubit", wires=10)

@qml.qnode(dev)
def qc():
    # GHZ state 
    qml.Hadamard(wires=0)
    for i in range(1,10):
        qml.CNOT(wires=[0,i])

    return qml.probs()
```

```python
qml.draw_mpl(qc)()
```

```python
# ghz com 2 qubits


dev = qml.device("default.qubit", wires=2)

@qml.qnode(dev)
def qc(angle):
    # GHZ state 
    qml.RY(angle, wires=0)
    for i in range(1,2):
        qml.CNOT(wires=[0,i])

    return qml.probs()
```

```python
angle = 2*np.arccos(3/5)
qc(angle)
```

```python
qml.draw_mpl(qc)(angle)
```

```python

dev = qml.device("default.qubit", wires=3)

@qml.qnode(dev)
def qc(angle):
    # GHZ state 
    qml.RY(angle, wires=0)
    qml.PauliZ(wires=0)
    qml.PhaseShift(np.pi/4, wires=0)

    qml.CNOT(wires=[0,1])
    qml.CNOT(wires=[0,2]) # igual  qml.CNOT(wires=[1,2]) tambem prepara GHZ
    return qml.state()
```

```python
angle = 2*np.arccos(2/7)
qml.draw_mpl(qc)(angle)
```
