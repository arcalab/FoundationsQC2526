```python
import pennylane as qml
```

```python
n_qubits = 1

dev = qml.device("default.qubit", wires=n_qubits)

@qml.qnode(dev)
def qc():
    #qubit começa sempre no estado |0> 
    #qml.PauliZ(wires=0) gate de pauli k aparece sempre como PauliK 
    qml.Hadamard(wires=0)

    #return qml.state()
    return qml.probs()



```

```python
qc()
```

```python
n_qubits = 2

dev = qml.device("default.qubit", wires=n_qubits)

@qml.qnode(dev)
def qc():
    #qubit começa sempre no estado |0> 
    #qml.PauliZ(wires=0) gate de pauli k aparece sempre como PauliK 
    qml.Hadamard(wires=0)
    qml.Hadamard(wires=1)
    qml.PauliZ(wires=0)
    qml.Hadamard(wires=0)

    #return qml.state()
    return qml.probs()



```

```python
qc()
```

```python
qml.draw_mpl(qc)()
```
